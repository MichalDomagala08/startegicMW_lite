#### Check Length of Files



```python
import os
import wave
import contextlib

# Set the directory containing the WAV files
foldname = 'Story4_AIsegments1' 
import os
import wave
import contextlib
import re
firstCount = 0;
secondCount = 0;
# Set the directory containing the WAV files
folder_path = f".\\"+foldname

# Regular expression to extract the numerical suffix
pattern = re.compile(r"segm_(\d+)\.wav")
firstEntityName  = "KAROLINA"
secondEntityName = "JANEK"


    
# Collect and sort files based on numeric suffix
wav_files = []
for filename in os.listdir(folder_path):
    match = pattern.match(filename)
    if match:
        wav_files.append((int(match.group(1)), filename))

# Sort files based on extracted numbers
wav_files.sort()
storyTimes = []
storyParts = []

# Process files in the correct order
for i, filename in enumerate(wav_files):
    file_path = os.path.join(folder_path, filename[1])

    if not i % 2:

        firstCount +=1
        partName = f"{firstEntityName}_{firstCount}"
    elif i % 2:
        secondCount +=1
        partName = f"{secondEntityName}_{secondCount}"

    with contextlib.closing(wave.open(file_path, 'r')) as wf:
        frames = wf.getnframes()
        rate = wf.getframerate()
        duration_ms = round((frames / rate) * 1000)/1000  # Convert to milliseconds
        storyTimes.append(duration_ms)
        storyParts.append(partName)
        print(f"{partName}: {duration_ms:.3f} ms")

storyDict = {}
storyDict['partNames'] = storyParts
storyDict['partTimes'] = storyTimes
```
